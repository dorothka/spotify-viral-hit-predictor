```python
import pandas as pd
import sqlite3

# 1. LOAD THE DATA
# We read the CSV file you just uploaded
df = pd.read_csv('spotify_hybrid.csv')

# 2. CREATE A VIRTUAL DATABASE
# We create a temporary SQL database in the computer's RAM
conn = sqlite3.connect(':memory:')
df.to_sql('tracks', conn, index=False, if_exists='replace')

# 3. THE ANALYTICAL QUERY (SQL)
# This is where we apply our business logic to find "Viral" characteristics
query = """
SELECT
    track_name,
    artists,
    album_name,
    (duration_ms / 60000.0) as duration_minutes,
    valence,
    energy,
    tempo,
    popularity,
    My_Rating,
    CASE
        WHEN valence > 0.6 AND energy > 0.6 THEN 'Party Anthem'
        WHEN valence < 0.4 AND energy < 0.4 THEN 'Sad/Moody'
        ELSE 'Chill/Casual'
    END as Vibe_Category
FROM
    tracks
WHERE
    duration_ms > 60000       -- Remove songs under 1 minute (usually glitches)
    AND popularity >= 1       -- Remove dead tracks with 0 popularity
"""

# 4. EXECUTE AND EXPORT
clean_df = pd.read_sql_query(query, conn)
clean_df.to_csv('final_spotify_data.csv', index=False)

print("SUCCESS! Processing complete.")
print(f"We analyzed {len(clean_df)} songs.")
print("Download 'final_spotify_data.csv' from the files folder on the left!")
```
